## 2. Carga del Dataset desde Hugging Face Hub

```python
from datasets import load_dataset
import pandas as pd

# Cargar dataset oficial desde Hugging Face Hub
ds = load_dataset('spacejourney/cdmx-sensory-impact-dataset')
df = pd.DataFrame(ds['train'])

print(f'✅ Se cargaron exitosamente {len(df)} santuarios sensoriales de CDMX.')
cols = ['name', 'category', 'neighborhood', 'sis_score', 'acoustic_noise_db', 'light_level_lux']
df[cols].head(10)
```

## 3. Visualización: Ruido Acústico vs. Sensory Impact Score (SIS)

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.set_theme(style='darkgrid')

scatter = sns.scatterplot(
    data=df,
    x='acoustic_noise_db',
    y='sis_score',
    hue='category',
    palette={'free': '#10B981', 'secret': '#E8C57D', 'premium': '#F59E0B'},
    s=150,
    edgecolor='black',
    linewidth=1.2
)

plt.title('CDMX: Presión Acústica (dBA) vs. Sensory Impact Score (SIS)', fontsize=14, weight='bold', pad=15)
plt.xlabel('Nivel Acústico (dBA IEC 61672-1) [Menor = Más Silencioso]', fontsize=11)
plt.ylabel('Sensory Impact Score (SIS) [0-100]', fontsize=11)
plt.axvline(x=55, color='red', linestyle='--', alpha=0.6, label='Umbral Confort OMS (55 dBA)')

for _, row in df.iterrows():
    if row['sis_score'] >= 92 or row['acoustic_noise_db'] <= 45:
        plt.annotate(
            row['name'],
            (row['acoustic_noise_db'], row['sis_score']),
            textcoords='offset points',
            xytext=(0, 10),
            ha='center',
            fontsize=8,
            weight='semibold'
        )

plt.legend(title='Categoría', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 4. Top 5 Santuarios de Mayor Calma Contemplativa (+1 de Hospitalidad)

```python
top_5 = df.sort_values(by='sis_score', ascending=False).head(5)

print('🏆 TOP 5 SANTUARIOS SENSORIALES DE CIUDAD DE MÉXICO:')
print('=' * 75)
for idx, (_, row) in enumerate(top_5.iterrows(), 1):
    print(f'\n{idx}. {row["name"]} (SIS: {row["sis_score"]}/100 | {row["acoustic_noise_db"]} dBA)')
    print(f'   Barrio: {row["neighborhood"]} | Categoría: {row["category"].upper()}')
    print(f'   Ventana Óptima: {row["golden_hour_window"]}')
    print(f'   💡 Consejo Omotenashi: {row["golden_spot_tips"]}')
```
